```python
import sys; sys.path.insert(0, "../src")
import numpy as np
import forecast as F
```

```python
df = F.load_all(); X = F.make_features(df); y = df["demand"]
ok = X.notna().all(axis=1); X, y = X[ok], y[ok]
train = X.index < F.TEST_START; test = ~train
feats = F.CALENDAR + F.LAGS + F.WEATHER
```

```python
def test_mae(features):
    m = F.fit(X.loc[train, features], y[train], 1455)
    return np.abs(y[test] - m.predict(X.loc[test, features])).mean()
```

```python

print("all features:", test_mae(feats))
for drop in [["solar_potential"], ["solar_potential", "solar_radiation"], ["wind_speed"]]:
    print("without", drop, test_mae([f for f in feats if f not in drop]))
```

```python
for drop in [["half_hour"], ["demand_lag_2d"], ["demand_lag_7d"], ["demand_lag_2d", "demand_lag_7d"]]:
    print("without", drop, test_mae([f for f in feats if f not in drop]))
```
